*Create an abstract Python class called Person that has the following attributes:<br>
○ Name
○ Age
○ Gender
○ Address<br>
● The class should implement inheritance<br>
● Define the magic str method that returns the basic info about the person<br>
● Define a method greet that accepts an instance of the Person class and greets the person<br>
e.g., the output should look like Hello John! My name is Jane.<br>
● Define an abstract method introduce that must be implemented by the child classes<br>
● Define a static method is_adult that accepts an argument age that returns True or False<br>
if the person is above 18 years old.
*

```python
from abc import ABC, abstractmethod
class Person:
    def __init__(self, name, age, gender, address):
        self.name = name
        self.age = age
        self.gender = gender
        self.address = address

    def __str__(self):
      return (f"Name: {self.name}, Age: {self.age}, Gender: {self.gender}, Address: {self.address}")

    def greet(self):
        print(f"Hello {self.name} My name is Jane ")


    @abstractmethod
    def introduce(self):
      pass

    @staticmethod
    def is_adult(age):
      if age > 18:
        return True
      else:
        return False

class child(Person):
    def __init__(self, name, age, gender, address,student_id):
      super().__init__(name, age, gender, address)
      self.student_id = student_id

    def introduce(self):
        return f"Hi, I'm {self.name}, a student with ID {self.student_id}."

p1=child("sim",21,"f","misrod",3)
p1.is_adult(21)
p1.greet()
p1.introduce()
```

Task 3: Multiple Inheritance, Polymorphism<br>
Create a Python class called Teacher that inherits from the Employee, Person classes
created
in Problem 1 and 2. This class must also implement cooperative inheritance
The Teacher class should have the following attributes:<br>
● Create a class attribute counter that will increase by one when a new instance of
employee is initialized and decrease by one when an instance is deleted<br>
● a private attribute teacher_id that holds the value according to counter e.g., TEC01,
TEC02. The private attribute must have only the getter method not setter, teacher_id
should not able to be changed once it is created<br>
● A subjects attribute is a list of subjects.
The class should have the following methods:<br>
● A constructor that initializes the attributes.<br>
● A method called counter wrapped in the property decorator that returns the class
variable counter<br>
● methods that appends or removes a particular subject from the subjects list.<br>
● An introduce method that overrides the abstract method defined in the Person class
that should return teacher_id and the list of subjects.<br>

● Since we have an attribute named teacher_id, we won’t need employee_id, override
the employee_id that now returns an AttributeError if someone tries to access the
attribute employee_id. E.g.Teacher object has no attribute employee id.<br>

```python
class Teacher(Employee):
    counter = 0

    def __init__(self, name, age, gender, address, salary, subjects=None):
        super().__init__(name, age, gender, address, salary)
        Teacher.counter += 1
        self.__teacher_id = f"TEC{Teacher.counter:02d}"
        self.subjects = subjects if subjects else []
    def __del__(self):
        Teacher.counter -= 1

    @property
    def teacher_id(self):
        return self.__teacher_id

    @property
    def get_counter(cls):
        return cls.counter

    def add_subject(self, subject):
        if subject not in self.subjects:
            self.subjects.append(subject)

    def remove_subject(self, subject):
        if subject in self.subjects:
            self.subjects.remove(subject)

    def introduce(self):
        return f"Hi, I'm {self.name}, a teacher with ID {self.teacher_id}. I teach: {', '.join(self.subjects) if self.subjects else 'No subjects assigned yet.'}"

    @property
    def employee_id(self):
        raise AttributeError("Teacher object has no attribute 'employee_id'")

teacher1 = Teacher("simran", 35, "F", "Indi", 70000, ["Math", "Science"])

print(teacher1.teacher_id)
print(teacher1.subjects)
print(teacher1.name)
print(teacher1.salary)
print(teacher1.introduce())
```
